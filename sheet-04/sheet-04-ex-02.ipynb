```python
import jax
import numpy as np


def deep_chain(w):
    return 0.5 * (np.prod(w) - 1) ** 2


print(jax.hessian(deep_chain)(np.array([1, 0.5, 0], dtype=np.float64)))
print(jax.hessian(deep_chain)(np.ones(3)))
```

```python
for l in range(1, 6):
    w = np.random.randn(l)
    hessian = jax.hessian(deep_chain)(w)
    print(hessian)
    print(np.mean(np.abs(hessian)))
```
