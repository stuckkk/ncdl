```python
import numpy as np

d = 100
n = 20

w_star = np.random.normal(size=(d,))
X = np.random.normal(size=(n, d))
y = X @ w_star

w = np.linalg.inv(X.T @ X) @ X.T @ y
hat_y = X @ w

print("loss = " + str(np.linalg.norm(hat_y - y) ** 2 / (2 * n)))
```

```python
w_new = X.T @ np.linalg.inv(X @ X.T) @ y
hat_y_new = X @ w_new
print("loss = " + str(np.linalg.norm(hat_y_new - y) ** 2 / (2 * n)))
```
